Imports

```python
import os
import time
from crud import NUUCloudClient
```

Create session

```python
client = NUUCloudClient()
client.session.cookies.set("session", 
                           # Replace with cookies whatever that means
                           "BUGGED", 
                           domain="nuu-cloud.replit.app")
try:
    client.login()
except Exception as e:
    print("Authentication failed with error :", e)
```

File reading

```python
TEST_MEDIA = 'test_files/test.png'
TEST_PDF = 'test_files/test.pdf'
TEST_MD = 'test_files/test.md'
```

Uploading

```python
# Uploading media
try:
    result = client.upload_media(TEST_MEDIA)
    print(result)
except Exception as e:
    print(f"Failed to upload media: {e}")
```

```python
# Uploading pdf
try:
    result = client.upload_document(TEST_PDF)
    print(result)
except Exception as e:
    print(f"Failed to upload pdf: {e}")
```

```python
# Uploading markdown
try:
    result = client.upload_document(TEST_MD)
    print(result)
except Exception as e:
    print(f"Failed to upload md: {e}")
```

Listing 

```python
ALL_MEDIA = {}
ALL_DOCUMENTS = {}
```

```python
# Listing media
try:
    media = client.list_media()
    print(f"Found {len(media)} media items:")
    for item in media:
        ALL_MEDIA[item['id']] = item
        print(f"   - {item['filename']} ({item['mimeType']}) - ID: {item['id']}")
except Exception as e:
    print(f"Failed to list media: {e}")
```

```python
# Listing documents (pdf + md)
try:
    documents = client.list_documents()
    print(f"Found {len(documents)} documents:")
    for doc in documents:
        ALL_DOCUMENTS[doc['id']] = doc
        print(f"   - {doc['title']} ({doc['contentType']}) - ID: {doc['id']}")
except Exception as e:
    print(f"Failed to list documents: {e}")
```

Downloading

```python
# Downloading media
try:
    client.download_media(list(ALL_MEDIA.keys())[0]['id'], 'downloaded_media.png')
except Exception as e:
    print(f"Failed to download media: {e}")
```

```python
# Downloading document
try:
    client.download_document(list(ALL_DOCUMENTS.keys())[0]['id'], 'downloaded_document.pdf')
except Exception as e:
    print(f"Failed to download document: {e}")
```

Deleting

```python
# Deleting media
try:
    client.delete_media(list(ALL_MEDIA.keys())[0]['id'])
except Exception as e:
    print(f"Failed to delete media: {e}")
```

```python
# Deleting document
try:
    client.delete_document(list(ALL_DOCUMENTS.keys())[0]['id'])
except Exception as e:
    print(f"Failed to delete document: {e}")
```
